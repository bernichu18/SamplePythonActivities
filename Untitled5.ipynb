```python
# Calculating the distance between points (x1,y1) and (x2,y2)
## Load the built-in math module.
import math

## Input all necessary variables.
x1 = float(input("enter x1"))
y1 = float(input("enter y1"))
x2 = float(input("enter x2"))
y2 = float(input("enter y2"))

## Input the distance formula.
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

## Print the final distance as your output.
print(f"\nThe distance between the two points is: {distance:2f}")
```
